```python
dff= dff[(dff['seller_pass_rate'] >= 0) & (dff['seller_pass_rate'] <= 100)]
```

```python
x= dff.drop('seller_pass_rate', axis=1)
y= dff['seller_pass_rate']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test =train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
scaler= StandardScaler()
x_train= scaler.fit_transform(x_train)
x_test= scaler.transform(x_test)
```

```python
from sklearn.linear_model import LinearRegression
lr= LinearRegression()
lr.fit(x_train, y_train)
```

```python
y_pred= lr.predict(x_test)
y_pred
```

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
import numpy as np
print("R2 Score:", r2_score(y_test, y_pred))
print("MAE:", mean_absolute_error(y_test, y_pred))
print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred)))
```

```python
y_pred_train= lr.predict(x_train)
print("Train R2 Score:", r2_score(y_train, y_pred_train))
```

```python
from sklearn.tree import DecisionTreeRegressor
dt = DecisionTreeRegressor(max_depth=10,min_samples_split=20,min_samples_leaf=10,random_state=42)
dt.fit(x_train,y_train)
```

```python
y_pred_dt= dt.predict(x_test)
```

```python
print("Decision Tree R2:",r2_score(y_test, y_pred_dt))
```

```python
y_train_pred_dt= dt.predict(x_train)
print("Train R2:",r2_score(y_train,y_train_pred_dt))
print("Test R2:",r2_score(y_test,y_pred_dt))
```

```python
if  r2_score(y_train, y_train_pred_dt)>r2_score(y_test, y_pred_dt)+.1:
    print('over fit')
elif r2_score(y_train, y_train_pred_dt)<.60:
    print('under fit')
else:
    print('good fit')
```

```python
from sklearn.ensemble import RandomForestRegressor
rf = RandomForestRegressor(n_estimators=50,max_depth=15,min_samples_split=10,min_samples_leaf=5,random_state=42,n_jobs=-1)
rf.fit(x_train, y_train)
y_pred_rf = rf.predict(x_test)
print("RF R2:",r2_score(y_test, y_pred_rf))
```

```python
y_train_pred_rf= rf.predict(x_train)
print("Train R2:",r2_score(y_train, y_train_pred_rf))
```

```python
if  r2_score(y_train,y_train_pred_rf)>r2_score(y_test,y_pred_rf)+.1:
    print('over fit')
elif r2_score(y_train,y_train_pred_rf)<.60:
    print('under fit')
else:
    print('good fit')
```

PLOTS

Sellers with more products sold generally show higher pass rates, indicating experience and reliability improve performance.

The scatter plot shows that sellers with more products sold tend to have higher and more consistent pass rates. New sellers exhibit high variability, while experienced sellers maintain stable performance. This indicates that seller experience is a strong indicator of seller reliability

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(7,5))
plt.scatter(dff['seller_products_sold'],dff['seller_pass_rate'],alpha=0.4,c='red')
plt.xlabel('Seller Products Sold')
plt.ylabel('Seller Pass Rate')
plt.title('Seller Experience vs Seller Pass Rate')
plt.xscale('log')
plt.show()
```

The visualization shows that seller experience has a significant impact on seller pass rate. New sellers have lower pass rates due to lack of platform familiarity, while experienced and top sellers consistently maintain high pass rates. This validates the importance of seller experience as a key predictive feature in seller performance models.

```python
import pandas as pd
import matplotlib.pyplot as plt
bins = [0, 10, 100, 1000, 10000, dff['seller_products_sold'].max()]
labels = ['New', 'Beginner', 'Intermediate', 'Experienced', 'Top Seller']
dff['experience_level'] = pd.cut(dff['seller_products_sold'],bins=bins,labels=labels)
avg_pass = dff.groupby('experience_level')['seller_pass_rate'].mean()
plt.figure(figsize=(8,5))
avg_pass.plot(kind='bar',color='pink')
plt.ylabel("Average Seller Pass Rate")
plt.xlabel("Seller Experience Level")
plt.title("Seller Experience vs Average Pass Rate")
plt.show()
```
